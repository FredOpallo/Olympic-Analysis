## Questions to be Answered using the olympic dataset

### QUESTION 1:Top 10 Athletes with the most medals won

```python
top_10_athletes=df.groupby(['Athlete','Country'])['Medal'].count().sort_values(ascending=False)
top_10_athletes.head(10)
```

### Question 2: Which Country won the most gold medal?

```python
country_gold_medals=df.query("Medal=='Gold'").groupby(['Country','Medal'])['Medal'].count().sort_values(ascending=False)
country_gold_medals
```

### Question 3: Which Country Won the Most Silver Medals?

```python
country_silver_medals=df.query("Medal=='Silver'").groupby(['Country','Medal'])['Medal'].count().sort_values(ascending=False)
country_silver_medals
```
